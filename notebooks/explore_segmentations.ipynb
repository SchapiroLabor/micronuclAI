```python
import matplotlib.pyplot as plt
from skimage.io import imsave, imread
from pathlib import Path
```

```python
# Set sample
sample = "Snap-2394"

# Set up Paths
input = Path(f"../data/training/ometif/{sample}.ome.tif").resolve()
unmicst = Path(f"../data/training/segmentation/unmicst/{sample}.tif").resolve()
mesmer_nuclear = Path(f"../data/training/segmentation/mesmer/{sample}.tif").resolve()
mesmer_whole = Path(f"../data/training/segmentation/mesmer_whole/{sample}.tif").resolve()
deepcell_nuclear = Path(f"../data/training/segmentation/deep_cell_nuclear/{sample}.tif").resolve()
deepcell_nuclear_preprocessed = Path(f"../data/training/segmentation/deep_cell_nuclear_preproces/{sample}.tif").resolve()
cellpose=Path(f"../data/training/segmentation/cellpose/{sample}.tif").resolve()
stardist=Path(f"../data/training/segmentation/stardist/{sample}.ome.tif").resolve()
out_path = Path("../figures/").resolve()
```

```python
# Read all the images
img = imread(input)
mask_unmicst = imread(unmicst)
mask_mesmer_nuclear = imread(mesmer_nuclear)
mask_mesmer_whole = imread(mesmer_whole)
mask_deepcell_nuclear = imread(deepcell_nuclear)
mask_deepcell_nuclear_preprocessed = imread(deepcell_nuclear_preprocessed)
mask_cellpose = imread(cellpose)
mask_stardist = imread(stardist)
```

```python
images = (img, mask_unmicst, mask_mesmer_nuclear, mask_mesmer_whole, mask_deepcell_nuclear, mask_deepcell_nuclear_preprocessed, mask_cellpose, mask_stardist)
names = ("dapi", "unmicst", "mesmer_nuclear", "mesmer_whole", "deepcell_nuclear", "deepcel_nuclear_preprocessed", "cellpose", "stardist")
```

```python
# Iterate over pairs of images/names
for i, name in zip(images, names):
    # Save figures with nice colors
    fig, ax = plt.subplots(1, 1, figsize=(14,19))
    ax.matshow(i)
    ax.set_xticks([])
    ax.set_yticks([])
    fig.savefig(out_path.joinpath(f"{sample}_{name}.png"), bbox_inches="tight", pad_inches=0)
```
