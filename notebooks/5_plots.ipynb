```python
import seaborn as sns
import matplotlib.pyplot as plt
from pathlib import Path
```

```python
# Lists with the values of cells and CIN_count
cells = [99, 100, 102, 102]
CIN_count = [11, 9, 10, 10]
names= ['manual', 'cellpose', 'mesmer', 'stardist']
```

```python
# Create a dataframe with the values
import pandas as pd
df = pd.DataFrame({'cells': cells, 'CIN count': CIN_count, 'segmentation': names})
```

```python
# Create a barplot for the CIN count
sns.set(style="white")
fig, ax = plt.subplots(figsize=(5, 4))
sns.barplot(y='segmentation', x='CIN count', data=df, palette="Blues_r", ax=ax)
ax.set_title("CIN counts manual vs micronuclAI", color="black", fontsize=20)
ax.set_ylabel("Segmentation", color="black", fontsize=20, fontweight='bold')
ax.set_xticklabels(ax.get_xticklabels(), fontsize=20, rotation=0)
ax.set_xlabel("CIN count", color="black", fontsize=20, fontweight='bold')
ax.set_yticklabels(ax.get_yticklabels(), fontsize=20)
sns.despine()
fig.savefig("../figures/BBBC039_segmentation_comparison.png", dpi=300, bbox_inches='tight', transparent=True)
fig.savefig("../figures/BBBC039_segmentation_comparison.pdf", dpi=300, bbox_inches='tight', transparent=True)
```

```python
# Create a barplot for the CIN count
sns.set(style="white")
fig, ax = plt.subplots(figsize=(5, 4))
sns.barplot(y='segmentation', x='cells', data=df, palette="Blues_r", ax=ax)
ax.set_title("Detected cells comparison", color="black", fontsize=20)
ax.set_ylabel("Segmentation", color="black", fontsize=20, fontweight='bold')
ax.set_xticklabels(ax.get_xticklabels(), fontsize=20, rotation=0)
ax.set_xlabel("Detected Cells", color="black", fontsize=20, fontweight='bold')
ax.set_yticklabels(ax.get_yticklabels(), fontsize=20)
sns.despine()
fig.savefig("../figures/BBBC039_segmentation_comparison_cells.png", dpi=300, bbox_inches='tight', transparent=True)
fig.savefig("../figures/BBBC039_segmentation_comparison_cells.pdf", dpi=300, bbox_inches='tight', transparent=True)
```
